```python
df = df.drop(columns=['Timestamp'])
```

```python
df.head()
```

```python
df.info()
```

```python
x = df.drop(columns=['Production Quality Score','Optimal Conditions'])
y = df['Production Quality Score']
```

```python
x.head()
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=1)
```

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
x_train.shape
```

```python
model = Sequential()
model.add(Dense(5,activation = 'relu',input_dim = 4))
model.add(Dense(10,activation = 'relu'))
model.add(Dense(15,activation='relu'))
model.add(Dense(20,activation='relu'))
model.add(Dense(1))
#
```

```python
model.summary()
```

```python
model.compile(loss='mean_squared_error', optimizer='Adam', metrics=['mean_absolute_error'])
```

```python
model.compile(loss='mean_squared_error', optimizer='Adam', metrics=['mean_absolute_error'])
model.fit(x_train,y_train,batch_size=10,epochs=10,validation_split=0.2)
```

```python
train_pred = model.predict(x_train)
# train_pred = np.where(train_pred>0.5,1,0)
train_pred
```

```python
from sklearn.metrics import mean_absolute_error
mean_absolute_error(y_train,train_pred)
```

```python
test_pred = model.predict(x_test)
from sklearn.metrics import mean_absolute_error
mean_absolute_error(y_test,test_pred)
```
