```python
from gtts import gTTS

text_to_say = "hello everyone.this is Vaibhav.i am good"

language = "en"

gtts_object = gTTS(text = text_to_say,
                  lang = language,
                   tld='co.in',
                  slow = True)

gtts_object.save("/content/gtts.wav")
```
