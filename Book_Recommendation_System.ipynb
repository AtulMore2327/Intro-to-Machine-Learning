```python
import pandas as pd
import numpy as np
import difflib
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
#
```

```python
df = pd.read_csv('/content/drive/MyDrive/BOOK/Books.csv')
```

```python
df.head()
```

```python
df.columns
```

```python
df = df.iloc[:10000,:]
```

```python
df.info()
```

```python
features = ['Book-Title','Book-Author','Year-Of-Publication','Publisher']
```

```python
for i in features:
  df[i] = df[i].fillna('')
```

```python
combine_features = df['Book-Title'].astype(str)+' '+df['Book-Author'].astype(str)+' '+df['Year-Of-Publication'].astype(str)+' '+df['Publisher'].astype(str)
```

```python
combine_features[0]
```

```python
vector = TfidfVectorizer()
```

```python
feature_vector = vector.fit_transform(combine_features)
```

```python
print(feature_vector)
```

```python
similarity = cosine_similarity(feature_vector)
```

```python
similarity.shape
```

```python
book_name = 'Decision in Normandy'
```

```python
list_of_all_title = df['Book-Title'].tolist()

print(list_of_all_title)
```

```python
find_close_match = difflib.get_close_matches(book_name, list_of_all_title)
print(find_close_match)
```

```python
close_match = find_close_match[0]
close_match
```

```python
index_of_book = df[df['Book-Title'] == close_match].index[0]
index_of_book
```

```python
similar_score = list(enumerate(similarity[index_of_book]))
print(similar_score)
```

```python
similar_score = sorted(similar_score,key = lambda x:x[1], reverse=True)
print(similar_score)
```

```python
df['Book-Title'][2]
```

```python
for i,j in similar_score[0:5]:
  print(df['Book-Title'][i])
```

```python
book_name = input('Enter your book name : ')
list_of_all_title = df['Book-Title'].tolist()
find_close_match = difflib.get_close_matches(book_name, list_of_all_title)
close_match = find_close_match[0]
index_of_book = df[df['Book-Title'] == close_match].index[0]
similar_score = list(enumerate(similarity[index_of_book]))
similar_score = sorted(similar_score,key = lambda x:x[1], reverse=True)
for i,j in similar_score[0:5]:
  print(df['Book-Title'][i])
```
