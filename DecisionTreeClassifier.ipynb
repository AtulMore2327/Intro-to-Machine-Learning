```python
import pandas as pd
from  sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv('/content/drive/MyDrive/Data  Set/Copy of daily_weather.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
x = df.drop(['high_humidity_3pm'],axis=1)
y = df['high_humidity_3pm']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model = DecisionTreeClassifier()
```

```python
model.fit(x_train,y_train)
```

```python
train_pred = model.predict(x_train)
test_pred = model.predict(x_test)
```

```python
print('Train ACC :-', accuracy_score(y_train, train_pred))
print('Test ACC :-', accuracy_score(y_test, test_pred))
```
